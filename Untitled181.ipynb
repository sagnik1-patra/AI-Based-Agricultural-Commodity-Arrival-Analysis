```python
import json
import warnings
from datetime import datetime, timezone
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import yaml

from sklearn.base import clone
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import (
    OneHotEncoder,
    StandardScaler
)
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import (
    train_test_split,
    KFold,
    cross_val_score
)
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings(
    "ignore",
    category=FutureWarning
)

# ==========================================================
# 1. PROJECT CONFIGURATION
# ==========================================================

PROJECT_DIR = Path(
    r"C:\Users\sagni\Downloads\AI-Based Agricultural Commodity Arrival Analysis"
)

DATASET_PATH = (
    PROJECT_DIR / "megamb_data_for_2012_n.xls"
)

PROJECT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

SEED = 42
TEST_SIZE = 0.20

rng = np.random.default_rng(SEED)

CATEGORICAL_FEATURES = [
    "District",
    "Block",
    "Market",
    "Month",
    "Commodity"
]

NUMERICAL_FEATURES = ["Year"]

FEATURE_COLUMNS = (
    CATEGORICAL_FEATURES + NUMERICAL_FEATURES
)

TARGET = "Arrival"

# AIS settings
POPULATION_SIZE = 8
GENERATIONS = 8
TOP_ANTIBODIES = 4
CLONES_PER_ANTIBODY = 3
RANDOM_NEW_ANTIBODIES = 2

CV_FOLDS = 3

sns.set_theme(style="whitegrid")

plt.rcParams["figure.figsize"] = (12, 7)
plt.rcParams["font.size"] = 11

# ==========================================================
# 2. LOAD AND CLEAN DATASET
# ==========================================================

print("\n========== LOADING DATASET ==========")

if not DATASET_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found: {DATASET_PATH}"
    )

df = pd.read_excel(
    DATASET_PATH,
    engine="xlrd"
)

df.columns = (
    df.columns.astype(str).str.strip()
)

df = df.dropna(how="all")
df = df.dropna(axis=1, how="all")

required = FEATURE_COLUMNS + [TARGET]

missing = [
    col for col in required
    if col not in df.columns
]

if missing:
    raise ValueError(
        f"Missing columns: {missing}"
    )

for col in CATEGORICAL_FEATURES:
    df[col] = (
        df[col]
        .astype("string")
        .str.strip()
        .fillna("Unknown")
    )

month_mapping = {
    "jan": "Jan",
    "feb": "Feb",
    "mar": "Mar",
    "apr": "Apr",
    "may": "May",
    "jun": "Jun",
    "jul": "Jul",
    "aug": "Aug",
    "sep": "Sep",
    "sept": "Sep",
    "oct": "Oct",
    "nov": "Nov",
    "dec": "Dec"
}

df["Month"] = (
    df["Month"]
    .str.lower()
    .replace(month_mapping)
)

df["Year"] = pd.to_numeric(
    df["Year"],
    errors="coerce"
)

df[TARGET] = pd.to_numeric(
    df[TARGET],
    errors="coerce"
)

df = df.dropna(subset=[TARGET])

df = df[
    df[TARGET] >= 0
].copy()

df = df.drop_duplicates()

if len(df) < 20:
    raise ValueError(
        "At least 20 valid records are recommended."
    )

print("Dataset Shape:", df.shape)
print(df.head())

df.to_csv(
    PROJECT_DIR / "ais_cleaned_dataset.csv",
    index=False
)

# ==========================================================
# 3. PREPARE TRAIN AND TEST DATA
# ==========================================================

X = df[FEATURE_COLUMNS].copy()

# Log transformation of arrival quantities
y = np.log1p(
    df[TARGET].astype(float)
)

X_train, X_test, y_train, y_test = (
    train_test_split(
        X,
        y,
        test_size=TEST_SIZE,
        random_state=SEED
    )
)

# Test data is not used during AIS optimization.

# ==========================================================
# 4. PREPROCESSING
# ==========================================================

categorical_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(
            strategy="most_frequent"
        )
    ),
    (
        "encoder",
        OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False
        )
    )
])

numerical_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(
            strategy="median"
        )
    ),
    (
        "scaler",
        StandardScaler()
    )
])

preprocessor = ColumnTransformer([
    (
        "categorical",
        categorical_pipeline,
        CATEGORICAL_FEATURES
    ),
    (
        "numerical",
        numerical_pipeline,
        NUMERICAL_FEATURES
    )
])

# ==========================================================
# 5. AIS ANTIBODY REPRESENTATION
# ==========================================================

# Each antibody represents a set of Random Forest
# hyperparameters:
#
# [n_estimators, max_depth,
#  min_samples_split, min_samples_leaf,
#  max_features_fraction]

BOUNDS = np.array([
    [50, 250],
    [3, 30],
    [2, 15],
    [1, 8],
    [0.3, 1.0]
], dtype=float)


def random_antibody():
    return rng.uniform(
        BOUNDS[:, 0],
        BOUNDS[:, 1]
    )


def decode_antibody(antibody):

    antibody = np.clip(
        antibody,
        BOUNDS[:, 0],
        BOUNDS[:, 1]
    )

    return {
        "n_estimators": int(
            round(antibody[0])
        ),
        "max_depth": int(
            round(antibody[1])
        ),
        "min_samples_split": int(
            round(antibody[2])
        ),
        "min_samples_leaf": int(
            round(antibody[3])
        ),
        "max_features": float(
            antibody[4]
        )
    }


def create_model(params):

    rf = RandomForestRegressor(
        **params,
        random_state=SEED,
        n_jobs=-1
    )

    return Pipeline([
        (
            "preprocessor",
            clone(preprocessor)
        ),
        (
            "regressor",
            rf
        )
    ])


# ==========================================================
# 6. FITNESS FUNCTION
# ==========================================================

# Minimize cross-validated RMSE in log-arrival space.
# A smaller fitness score is better.

fitness_cache = {}

cv = KFold(
    n_splits=CV_FOLDS,
    shuffle=True,
    random_state=SEED
)


def fitness(antibody):

    params = decode_antibody(antibody)

    key = (
        params["n_estimators"],
        params["max_depth"],
        params["min_samples_split"],
        params["min_samples_leaf"],
        round(params["max_features"], 4)
    )

    if key in fitness_cache:
        return fitness_cache[key]

    model = create_model(params)

    scores = cross_val_score(
        model,
        X_train,
        y_train,
        cv=cv,
        scoring="neg_mean_squared_error",
        n_jobs=1
    )

    rmse = float(
        np.sqrt(-scores.mean())
    )

    fitness_cache[key] = rmse

    return rmse


# ==========================================================
# 7. CLONAL SELECTION / MUTATION
# ==========================================================

def mutate_antibody(
    antibody,
    mutation_strength
):

    span = (
        BOUNDS[:, 1] - BOUNDS[:, 0]
    )

    noise = rng.normal(
        loc=0.0,
        scale=mutation_strength,
        size=len(antibody)
    )

    mutated = (
        antibody + noise * span
    )

    return np.clip(
        mutated,
        BOUNDS[:, 0],
        BOUNDS[:, 1]
    )


def run_ais():

    print("\n========== AIS OPTIMIZATION ==========")

    population = [
        random_antibody()
        for _ in range(POPULATION_SIZE)
    ]

    global_best = None
    global_best_score = float("inf")

    history = []

    for generation in range(
        1, GENERATIONS + 1
    ):

        scored = [
            (fitness(a), a.copy())
            for a in population
        ]

        scored.sort(
            key=lambda item: item[0]
        )

        generation_best_score = scored[0][0]
        generation_best = scored[0][1].copy()

        if generation_best_score < global_best_score:
            global_best_score = generation_best_score
            global_best = generation_best.copy()

        print(
            f"Generation {generation}/{GENERATIONS}"
            f" | Best CV Log-RMSE: "
            f"{global_best_score:.5f}"
        )

        history.append({
            "Generation": generation,
            "Generation_Best_Log_RMSE":
                generation_best_score,
            "Global_Best_Log_RMSE":
                global_best_score,
            "Evaluated_Configurations":
                len(fitness_cache)
        })

        selected = scored[:TOP_ANTIBODIES]

        clones = []

        for rank, (score, antibody) in enumerate(
            selected
        ):

            # Better antibodies receive smaller
            # mutations; weaker ones explore more.

            mutation_strength = (
                0.03 + 0.05 * rank
            )

            for _ in range(
                CLONES_PER_ANTIBODY
            ):

                clone_antibody = mutate_antibody(
                    antibody,
                    mutation_strength
                )

                clones.append(
                    clone_antibody
                )

        candidates = (
            [item[1] for item in selected]
            + clones
            + [
                random_antibody()
                for _ in range(
                    RANDOM_NEW_ANTIBODIES
                )
            ]
        )

        candidate_scores = [
            (fitness(a), a.copy())
            for a in candidates
        ]

        candidate_scores.sort(
            key=lambda item: item[0]
        )

        population = [
            item[1]
            for item in candidate_scores[
                :POPULATION_SIZE
            ]
        ]

    # Include the final population in best search.
    final_scores = [
        (fitness(a), a.copy())
        for a in population
    ]

    final_scores.sort(
        key=lambda item: item[0]
    )

    if final_scores[0][0] < global_best_score:
        global_best_score = final_scores[0][0]
        global_best = final_scores[0][1].copy()

    return (
        decode_antibody(global_best),
        global_best_score,
        pd.DataFrame(history)
    )


best_params, best_cv_score, history_df = run_ais()

print("\nBest AIS Hyperparameters:")
print(best_params)

print("\nBest CV Log-RMSE:")
print(best_cv_score)

history_df.to_csv(
    PROJECT_DIR / "ais_optimization_history.csv",
    index=False
)

# ==========================================================
# 8. TRAIN OPTIMIZED MODEL
# ==========================================================

print("\n========== TRAINING AIS MODEL ==========")

ais_model = create_model(
    best_params
)

ais_model.fit(
    X_train,
    y_train
)

joblib.dump(
    ais_model,
    PROJECT_DIR / "ais_agricultural_model.pkl"
)

print("Saved: ais_agricultural_model.pkl")

# ==========================================================
# 9. EVALUATE ON HELD-OUT TEST DATA
# ==========================================================

def inverse_target(values):

    values = np.asarray(
        values
    ).flatten()

    return np.maximum(
        np.expm1(
            np.clip(values, -20, 20)
        ),
        0
    )


def evaluate(actual, predicted):

    return {
        "MAE": float(
            mean_absolute_error(
                actual, predicted
            )
        ),
        "RMSE": float(
            np.sqrt(
                mean_squared_error(
                    actual, predicted
                )
            )
        ),
        "R2": float(
            r2_score(
                actual, predicted
            )
        )
    }


actual = inverse_target(
    y_test.to_numpy()
)

predicted = inverse_target(
    ais_model.predict(X_test)
)

ais_metrics = evaluate(
    actual,
    predicted
)

print("\n========== AIS TEST METRICS ==========")

for name, value in ais_metrics.items():
    print(f"{name}: {value:.5f}")

metrics_df = pd.DataFrame([{
    "Model": "AIS Optimized Random Forest",
    **ais_metrics
}])

metrics_df.to_csv(
    PROJECT_DIR / "ais_evaluation_metrics.csv",
    index=False
)

# ==========================================================
# 10. SAVE RESULT CSV
# ==========================================================

result_df = X_test.copy()

result_df["Actual_Arrival"] = actual

result_df["AIS_Predicted_Arrival"] = predicted

result_df["Absolute_Error"] = np.abs(
    actual - predicted
)

result_df["Squared_Error"] = (
    actual - predicted
) ** 2

result_df.to_csv(
    PROJECT_DIR / "ais_result.csv",
    index=False
)

print("Saved: ais_result.csv")

# ==========================================================
# 11. AIS ACCURACY GRAPH
# ==========================================================

fig, axes = plt.subplots(
    1, 3,
    figsize=(16, 5)
)

for ax, metric in zip(
    axes,
    ["R2", "MAE", "RMSE"]
):

    value = ais_metrics[metric]

    ax.bar(
        ["AIS Random Forest"],
        [value]
    )

    ax.set_title(
        f"AIS {metric}",
        fontweight="bold"
    )

    ax.set_ylabel(metric)

    ax.text(
        0,
        value,
        f"{value:.4f}",
        ha="center",
        va="bottom"
        if value >= 0 else "top"
    )

fig.suptitle(
    "AIS-Optimized Agricultural Arrival Model Performance",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "ais_accuracy_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ==========================================================
# 12. AIS HEATMAP
# ==========================================================

top_commodities = (
    df.groupby("Commodity")["Arrival"]
    .sum()
    .nlargest(15)
    .index
)

top_markets = (
    df.groupby("Market")["Arrival"]
    .sum()
    .nlargest(12)
    .index
)

heatmap_df = df[
    df["Commodity"].isin(top_commodities)
    & df["Market"].isin(top_markets)
]

pivot = heatmap_df.pivot_table(
    index="Commodity",
    columns="Market",
    values="Arrival",
    aggfunc="sum",
    fill_value=0
)

plt.figure(figsize=(16, 10))

sns.heatmap(
    pivot,
    annot=True,
    fmt=".0f",
    cmap="YlGnBu",
    linewidths=0.4,
    cbar_kws={
        "label": "Arrival Quantity (Quintals)"
    }
)

plt.title(
    "AIS Project - Agricultural Commodity Arrival Heatmap",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Market")
plt.ylabel("Commodity")

plt.xticks(
    rotation=45,
    ha="right"
)

plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "ais_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ==========================================================
# 13. BASELINE VS AIS COMPARISON
# ==========================================================

comparison_rows = [{
    "Model": "AIS Optimized Random Forest",
    **ais_metrics
}]

baseline_model_path = (
    PROJECT_DIR / "agricultural_model.pkl"
)

if baseline_model_path.exists():

    baseline_model = joblib.load(
        baseline_model_path
    )

    baseline_pred = inverse_target(
        baseline_model.predict(X_test)
    )

    baseline_metrics = evaluate(
        actual,
        baseline_pred
    )

    comparison_rows.append({
        "Model": "Baseline Best ML",
        **baseline_metrics
    })

    print("\nBaseline Model Metrics:")
    print(baseline_metrics)

else:
    print(
        "\nBaseline model not found. "
        "AIS-only comparison will be generated."
    )

comparison_df = pd.DataFrame(
    comparison_rows
)

comparison_df.to_csv(
    PROJECT_DIR / "ais_comparison.csv",
    index=False
)

fig, axes = plt.subplots(
    1, 3,
    figsize=(18, 6)
)

for ax, metric in zip(
    axes,
    ["MAE", "RMSE", "R2"]
):

    sns.barplot(
        data=comparison_df,
        x=metric,
        y="Model",
        ax=ax
    )

    ax.set_title(
        f"{metric} Comparison",
        fontweight="bold"
    )

    ax.set_ylabel("")

fig.suptitle(
    "Baseline vs AIS-Optimized Model",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "ais_comparison_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ==========================================================
# 14. AIS RESULT GRAPH
# ==========================================================

plot_df = result_df.head(
    min(100, len(result_df))
).reset_index(drop=True)

plt.figure(figsize=(16, 7))

plt.plot(
    plot_df.index,
    plot_df["Actual_Arrival"],
    marker="o",
    markersize=3,
    label="Actual Arrival"
)

plt.plot(
    plot_df.index,
    plot_df["AIS_Predicted_Arrival"],
    marker="s",
    markersize=3,
    label="AIS Predicted Arrival"
)

plt.title(
    "AIS Actual vs Predicted Commodity Arrivals",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Test Sample Index")
plt.ylabel("Arrival (Quintals)")

plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "ais_result_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ==========================================================
# 15. AIS PREDICTION CSV
# ==========================================================

# Predictions on all dataset records.
# Includes training records and therefore should
# not be used for independent accuracy evaluation.

all_features = df[FEATURE_COLUMNS].copy()

all_predictions = inverse_target(
    ais_model.predict(all_features)
)

prediction_df = all_features.copy()

prediction_df["Actual_Arrival"] = (
    df["Arrival"].to_numpy()
)

prediction_df["AIS_Predicted_Arrival"] = (
    all_predictions
)

prediction_df["Absolute_Error"] = np.abs(
    prediction_df["Actual_Arrival"]
    - prediction_df["AIS_Predicted_Arrival"]
)

prediction_df.to_csv(
    PROJECT_DIR / "ais_prediction.csv",
    index=False
)

print("Saved: ais_prediction.csv")

# ==========================================================
# 16. AIS PREDICTION GRAPH
# ==========================================================

commodity_prediction = (
    prediction_df.groupby(
        "Commodity",
        as_index=False
    )[
        [
            "Actual_Arrival",
            "AIS_Predicted_Arrival"
        ]
    ]
    .sum()
    .sort_values(
        "Actual_Arrival",
        ascending=False
    )
    .head(15)
    .set_index("Commodity")
)

commodity_prediction.plot(
    kind="bar",
    figsize=(16, 7),
    width=0.8
)

plt.title(
    "AIS Commodity-wise Actual vs Predicted Arrivals",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Commodity")
plt.ylabel("Total Arrival (Quintals)")

plt.xticks(
    rotation=45,
    ha="right"
)

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "ais_prediction_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ==========================================================
# 17. AIS OPTIMIZATION CONVERGENCE GRAPH
# ==========================================================

plt.figure(figsize=(12, 6))

plt.plot(
    history_df["Generation"],
    history_df["Global_Best_Log_RMSE"],
    marker="o",
    linewidth=2,
    label="Best AIS Fitness"
)

plt.plot(
    history_df["Generation"],
    history_df["Generation_Best_Log_RMSE"],
    linestyle="--",
    marker="s",
    label="Generation Best"
)

plt.title(
    "AIS Clonal Selection Optimization Convergence",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Generation")
plt.ylabel("Cross-Validated Log-RMSE")

plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "ais_optimization_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ==========================================================
# 18. SAVE AIS YAML CONFIGURATION
# ==========================================================

config = {
    "project_name":
        "AI-Based Agricultural Commodity Arrival Analysis",

    "optimization_algorithm":
        "Artificial Immune System - Clonal Selection",

    "dataset": str(DATASET_PATH),

    "target": TARGET,

    "target_unit": "Quintals",

    "features": FEATURE_COLUMNS,

    "random_state": SEED,

    "test_size": TEST_SIZE,

    "population_size": POPULATION_SIZE,

    "generations": GENERATIONS,

    "top_antibodies": TOP_ANTIBODIES,

    "clones_per_antibody": CLONES_PER_ANTIBODY,

    "random_new_antibodies":
        RANDOM_NEW_ANTIBODIES,

    "cross_validation_folds": CV_FOLDS,

    "fitness_metric":
        "Cross-validated log-space RMSE",

    "best_hyperparameters": best_params,

    "best_cv_log_rmse": float(
        best_cv_score
    ),

    "model_file":
        "ais_agricultural_model.pkl"
}

with open(
    PROJECT_DIR / "ais_config.yaml",
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        config,
        file,
        sort_keys=False
    )

# ==========================================================
# 19. SAVE AIS METADATA JSON
# ==========================================================

metadata = {
    "project_name":
        "AI-Based Agricultural Commodity Arrival Analysis",

    "algorithm":
        "AIS Clonal Selection",

    "created_at":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "dataset_records": int(
        len(df)
    ),

    "training_records": int(
        len(X_train)
    ),

    "testing_records": int(
        len(X_test)
    ),

    "features": FEATURE_COLUMNS,

    "target": TARGET,

    "target_unit": "Quintals",

    "best_hyperparameters":
        best_params,

    "best_cv_log_rmse": float(
        best_cv_score
    ),

    "test_metrics": ais_metrics,

    "comparison_results":
        comparison_df.to_dict(
            orient="records"
        ),

    "evaluation_method":
        "Held-out random 80/20 split",

    "note":
        "AIS optimizes Random Forest parameters "
        "using training-set cross-validation. "
        "The test set is used for final evaluation."
}

with open(
    PROJECT_DIR / "ais_model_metadata.json",
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        metadata,
        file,
        indent=4
    )

# ==========================================================
# 20. FINAL OUTPUT VERIFICATION
# ==========================================================

print("\n========== AIS OUTPUT FILES ==========")

output_files = [
    "ais_agricultural_model.pkl",
    "ais_config.yaml",
    "ais_model_metadata.json",
    "ais_cleaned_dataset.csv",
    "ais_accuracy_graph.png",
    "ais_heatmap.png",
    "ais_comparison.csv",
    "ais_comparison_graph.png",
    "ais_result.csv",
    "ais_result_graph.png",
    "ais_prediction.csv",
    "ais_prediction_graph.png",
    "ais_optimization_graph.png",
    "ais_optimization_history.csv",
    "ais_evaluation_metrics.csv"
]

for filename in output_files:

    path = PROJECT_DIR / filename

    if path.exists():
        print("[SAVED]", filename)
    else:
        print("[MISSING]", filename)

print("\n========================================")
print("AIS OPTIMIZATION COMPLETED SUCCESSFULLY")
print("========================================")

print("\nBest Parameters:")
print(best_params)

print("\nAIS Test Metrics:")
print(ais_metrics)

print("\nSaved Directory:")
print(PROJECT_DIR)
```
