```python

import os
import json
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# ============================================================
# 1. PROJECT CONFIGURATION
# ============================================================

PROJECT_DIR = Path(
    r"C:\Users\sagni\Downloads\AI-Based Agricultural Commodity Arrival Analysis"
)

DATASET_PATH = PROJECT_DIR / "megamb_data_for_2012_n.xls"
PKL_MODEL_PATH = PROJECT_DIR / "agricultural_model.pkl"
H5_MODEL_PATH = PROJECT_DIR / "agricultural_model.h5"
PREPROCESSOR_PATH = PROJECT_DIR / "preprocessor.pkl"
METADATA_PATH = PROJECT_DIR / "model_metadata.json"
COMPARISON_PATH = PROJECT_DIR / "model_comparison.csv"

PROJECT_DIR.mkdir(parents=True, exist_ok=True)

RANDOM_STATE = 42
TEST_SIZE = 0.20

FEATURE_COLUMNS = [
    "District",
    "Block",
    "Market",
    "Month",
    "Commodity",
    "Year"
]

CATEGORICAL_COLUMNS = [
    "District",
    "Block",
    "Market",
    "Month",
    "Commodity"
]

TARGET_COLUMN = "Arrival"

sns.set_theme(style="whitegrid")

plt.rcParams["figure.figsize"] = (12, 7)
plt.rcParams["font.size"] = 11

# ============================================================
# 2. LOAD DATASET
# ============================================================

print("\n========== LOADING DATASET ==========")

df = pd.read_excel(
    DATASET_PATH,
    engine="xlrd"
)

df.columns = df.columns.astype(str).str.strip()

df = df.dropna(how="all")
df = df.dropna(axis=1, how="all")

# ============================================================
# 3. CLEAN DATA
# Must match the original training script
# ============================================================

for col in CATEGORICAL_COLUMNS:
    df[col] = (
        df[col]
        .astype("string")
        .str.strip()
        .fillna("Unknown")
    )

month_mapping = {
    "jan": "Jan",
    "feb": "Feb",
    "mar": "Mar",
    "apr": "Apr",
    "may": "May",
    "jun": "Jun",
    "jul": "Jul",
    "aug": "Aug",
    "sep": "Sep",
    "sept": "Sep",
    "oct": "Oct",
    "nov": "Nov",
    "dec": "Dec"
}

df["Month"] = (
    df["Month"]
    .str.lower()
    .replace(month_mapping)
)

df["Year"] = pd.to_numeric(
    df["Year"],
    errors="coerce"
)

df["Arrival"] = pd.to_numeric(
    df["Arrival"],
    errors="coerce"
)

df = df.dropna(subset=["Arrival"])
df = df[df["Arrival"] >= 0].copy()
df = df.drop_duplicates()

print("Dataset Shape:", df.shape)
print(df.head())

# ============================================================
# 4. LOAD SAVED MODELS
# ============================================================

print("\n========== LOADING MODELS ==========")

required_files = [
    PKL_MODEL_PATH,
    H5_MODEL_PATH,
    PREPROCESSOR_PATH
]

for file_path in required_files:
    if not file_path.exists():
        raise FileNotFoundError(
            f"Required file missing: {file_path}\n"
            "Run train_agricultural_model.py first."
        )

ml_model = joblib.load(PKL_MODEL_PATH)

nn_model = tf.keras.models.load_model(
    str(H5_MODEL_PATH),
    compile=False
)

nn_preprocessor = joblib.load(
    PREPROCESSOR_PATH
)

print("Machine Learning Model Loaded")
print("Neural Network Model Loaded")
print("Preprocessor Loaded")

# ============================================================
# 5. RECREATE ORIGINAL TRAIN / TEST SPLIT
# ============================================================

X = df[FEATURE_COLUMNS].copy()

y = np.log1p(
    df[TARGET_COLUMN].astype(float)
)

X_train, X_test, y_train, y_test = (
    train_test_split(
        X,
        y,
        test_size=TEST_SIZE,
        random_state=RANDOM_STATE
    )
)

actual = np.expm1(y_test.to_numpy())

# ============================================================
# 6. GENERATE PREDICTIONS
# ============================================================

print("\n========== GENERATING PREDICTIONS ==========")

def inverse_target(values):
    values = np.asarray(values).flatten()

    return np.maximum(
        np.expm1(
            np.clip(values, -20, 20)
        ),
        0
    )


# Machine learning predictions
ml_predictions_log = ml_model.predict(X_test)

ml_predictions = inverse_target(
    ml_predictions_log
)

# Neural network predictions
X_test_nn = nn_preprocessor.transform(
    X_test
).astype("float32")

nn_predictions_log = nn_model.predict(
    X_test_nn,
    verbose=0
).flatten()

nn_predictions = inverse_target(
    nn_predictions_log
)

# ============================================================
# 7. EVALUATION METRICS
# ============================================================

def calculate_metrics(actual, predicted):
    return {
        "MAE": float(
            mean_absolute_error(actual, predicted)
        ),
        "RMSE": float(
            np.sqrt(
                mean_squared_error(
                    actual, predicted
                )
            )
        ),
        "R2": float(
            r2_score(actual, predicted)
        )
    }


ml_metrics = calculate_metrics(
    actual,
    ml_predictions
)

nn_metrics = calculate_metrics(
    actual,
    nn_predictions
)

print("\nMachine Learning Metrics:")
print(ml_metrics)

print("\nNeural Network Metrics:")
print(nn_metrics)

# ============================================================
# 8. SAVE RESULT CSV
# ============================================================

print("\n========== SAVING RESULT CSV ==========")

result_df = X_test.copy()

result_df["Actual_Arrival"] = actual

result_df["ML_Predicted_Arrival"] = (
    ml_predictions
)

result_df["NN_Predicted_Arrival"] = (
    nn_predictions
)

result_df["ML_Absolute_Error"] = np.abs(
    actual - ml_predictions
)

result_df["NN_Absolute_Error"] = np.abs(
    actual - nn_predictions
)

result_df["ML_Squared_Error"] = (
    actual - ml_predictions
) ** 2

result_df["NN_Squared_Error"] = (
    actual - nn_predictions
) ** 2

result_df.to_csv(
    PROJECT_DIR / "result.csv",
    index=False
)

print("Saved: result.csv")

# ============================================================
# 9. ACCURACY GRAPH
# R2, MAE and RMSE
# ============================================================

print("\n========== GENERATING ACCURACY GRAPH ==========")

metric_df = pd.DataFrame([
    {
        "Model": "Best ML Model",
        **ml_metrics
    },
    {
        "Model": "Neural Network",
        **nn_metrics
    }
])

fig, axes = plt.subplots(
    1, 3,
    figsize=(17, 5)
)

metric_names = ["R2", "MAE", "RMSE"]

for ax, metric in zip(axes, metric_names):

    sns.barplot(
        data=metric_df,
        x="Model",
        y=metric,
        ax=ax
    )

    ax.set_title(
        f"{metric} Comparison",
        fontsize=13,
        fontweight="bold"
    )

    ax.set_xlabel("Model")
    ax.set_ylabel(metric)

    for container in ax.containers:
        ax.bar_label(
            container,
            fmt="%.3f"
        )

fig.suptitle(
    "Agricultural Commodity Arrival Prediction Performance",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "accuracy_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

print("Saved: accuracy_graph.png")

# ============================================================
# 10. HEATMAP
# Commodity vs Market Arrival Quantity
# ============================================================

print("\n========== GENERATING HEATMAP ==========")

# Select top commodities and markets by total arrivals

top_commodities = (
    df.groupby("Commodity")["Arrival"]
    .sum()
    .nlargest(15)
    .index
)

top_markets = (
    df.groupby("Market")["Arrival"]
    .sum()
    .nlargest(12)
    .index
)

heatmap_data = df[
    df["Commodity"].isin(top_commodities)
    & df["Market"].isin(top_markets)
]

pivot_table = heatmap_data.pivot_table(
    index="Commodity",
    columns="Market",
    values="Arrival",
    aggfunc="sum",
    fill_value=0
)

plt.figure(figsize=(16, 10))

sns.heatmap(
    pivot_table,
    cmap="YlGnBu",
    annot=True,
    fmt=".0f",
    linewidths=0.4,
    cbar_kws={
        "label": "Total Arrival (Quintals)"
    }
)

plt.title(
    "Agricultural Commodity Arrival Heatmap",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Market")
plt.ylabel("Commodity")

plt.xticks(rotation=45, ha="right")
plt.yticks(rotation=0)

plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

print("Saved: heatmap.png")

# ============================================================
# 11. MODEL COMPARISON GRAPH
# ============================================================

print("\n========== GENERATING COMPARISON GRAPH ==========")

if COMPARISON_PATH.exists():

    comparison_df = pd.read_csv(
        COMPARISON_PATH
    )

else:

    comparison_df = metric_df.copy()

comparison_df = comparison_df.dropna(
    subset=["Model", "MAE", "RMSE", "R2"]
)

fig, axes = plt.subplots(
    1, 3,
    figsize=(19, 6)
)

for ax, metric in zip(
    axes,
    ["MAE", "RMSE", "R2"]
):

    ordered_df = comparison_df.sort_values(
        metric,
        ascending=(metric != "R2")
    )

    sns.barplot(
        data=ordered_df,
        x=metric,
        y="Model",
        ax=ax
    )

    ax.set_title(
        f"Model Comparison - {metric}",
        fontweight="bold"
    )

    ax.set_xlabel(metric)
    ax.set_ylabel("")

fig.suptitle(
    "Machine Learning and Deep Learning Model Comparison",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "comparison_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

print("Saved: comparison_graph.png")

# ============================================================
# 12. RESULT GRAPH
# Actual vs Predicted Arrivals
# ============================================================

print("\n========== GENERATING RESULT GRAPH ==========")

# Show up to 100 test samples for readability
display_count = min(100, len(result_df))

plot_df = result_df.iloc[
    :display_count
].reset_index(drop=True)

plt.figure(figsize=(16, 7))

plt.plot(
    plot_df.index,
    plot_df["Actual_Arrival"],
    marker="o",
    markersize=3,
    label="Actual Arrival"
)

plt.plot(
    plot_df.index,
    plot_df["ML_Predicted_Arrival"],
    marker="s",
    markersize=3,
    label="ML Prediction"
)

plt.plot(
    plot_df.index,
    plot_df["NN_Predicted_Arrival"],
    marker="^",
    markersize=3,
    label="Neural Network Prediction"
)

plt.title(
    "Actual vs Predicted Agricultural Commodity Arrivals",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Test Sample Index")
plt.ylabel("Arrival Quantity (Quintals)")

plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "result_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

print("Saved: result_graph.png")

# ============================================================
# 13. GENERATE PREDICTIONS FOR DATASET
# ============================================================

print("\n========== GENERATING PREDICTION CSV ==========")

# These are fitted-model estimates for existing
# dataset rows, not future-year forecasts.

all_features = df[FEATURE_COLUMNS].copy()

all_ml_log = ml_model.predict(
    all_features
)

all_ml_predictions = inverse_target(
    all_ml_log
)

all_nn_features = nn_preprocessor.transform(
    all_features
).astype("float32")

all_nn_log = nn_model.predict(
    all_nn_features,
    verbose=0
).flatten()

all_nn_predictions = inverse_target(
    all_nn_log
)

prediction_df = all_features.copy()

prediction_df["Actual_Arrival"] = (
    df["Arrival"].to_numpy()
)

prediction_df["ML_Predicted_Arrival"] = (
    all_ml_predictions
)

prediction_df["NN_Predicted_Arrival"] = (
    all_nn_predictions
)

prediction_df["ML_Absolute_Error"] = np.abs(
    prediction_df["Actual_Arrival"]
    - prediction_df["ML_Predicted_Arrival"]
)

prediction_df["NN_Absolute_Error"] = np.abs(
    prediction_df["Actual_Arrival"]
    - prediction_df["NN_Predicted_Arrival"]
)

prediction_df.to_csv(
    PROJECT_DIR / "prediction.csv",
    index=False
)

print("Saved: prediction.csv")

# ============================================================
# 14. PREDICTION GRAPH
# Commodity-wise Predicted Arrivals
# ============================================================

print("\n========== GENERATING PREDICTION GRAPH ==========")

commodity_predictions = (
    prediction_df.groupby(
        "Commodity",
        as_index=False
    )[
        [
            "Actual_Arrival",
            "ML_Predicted_Arrival",
            "NN_Predicted_Arrival"
        ]
    ]
    .sum()
)

commodity_predictions = (
    commodity_predictions.sort_values(
        "Actual_Arrival",
        ascending=False
    ).head(15)
)

commodity_predictions = (
    commodity_predictions.set_index("Commodity")
)

ax = commodity_predictions.plot(
    kind="bar",
    figsize=(17, 8),
    width=0.8
)

plt.title(
    "Commodity-wise Actual and Predicted Arrivals",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Commodity")
plt.ylabel("Total Arrival Quantity (Quintals)")

plt.xticks(rotation=45, ha="right")
plt.legend(title="Arrival Type")
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "prediction_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

print("Saved: prediction_graph.png")

# ============================================================
# 15. ADDITIONAL METRICS CSV
# ============================================================

metric_df.to_csv(
    PROJECT_DIR / "evaluation_metrics.csv",
    index=False
)

# ============================================================
# 16. VERIFY OUTPUT FILES
# ============================================================

print("\n========== VERIFYING OUTPUTS ==========")

output_files = [
    "accuracy_graph.png",
    "heatmap.png",
    "comparison_graph.png",
    "result.csv",
    "result_graph.png",
    "prediction.csv",
    "prediction_graph.png",
    "evaluation_metrics.csv"
]

for filename in output_files:

    output_path = PROJECT_DIR / filename

    if output_path.exists():
        print("[SAVED]", filename)
    else:
        print("[MISSING]", filename)

print("\n============================================")
print("ALL VISUALIZATIONS GENERATED SUCCESSFULLY")
print("============================================")

print("\nMachine Learning Results:")
print(ml_metrics)

print("\nNeural Network Results:")
print(nn_metrics)

print("\nOutput Directory:")
print(PROJECT_DIR)
```
