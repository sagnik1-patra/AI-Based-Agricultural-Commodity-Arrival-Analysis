```python

import os
import json
import pickle
import warnings
from datetime import datetime, timezone
from pathlib import Path

import numpy as np
import pandas as pd
import yaml
import joblib

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split
from sklearn.ensemble import (
    RandomForestRegressor,
    GradientBoostingRegressor,
    ExtraTreesRegressor
)
from sklearn.linear_model import Ridge
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
from sklearn.base import clone

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

warnings.filterwarnings("ignore", category=FutureWarning)

# =====================================================
# 1. PROJECT CONFIGURATION
# =====================================================

PROJECT_DIR = Path(
    r"C:\Users\sagni\Downloads\AI-Based Agricultural Commodity Arrival Analysis"
)

DATASET_PATH = PROJECT_DIR / "megamb_data_for_2012_n.xls"

PROJECT_DIR.mkdir(parents=True, exist_ok=True)

RANDOM_STATE = 42
TEST_SIZE = 0.20

np.random.seed(RANDOM_STATE)
tf.keras.utils.set_random_seed(RANDOM_STATE)

CATEGORICAL_FEATURES = [
    "District",
    "Block",
    "Market",
    "Month",
    "Commodity"
]

NUMERICAL_FEATURES = ["Year"]

FEATURE_COLUMNS = (
    CATEGORICAL_FEATURES + NUMERICAL_FEATURES
)

TARGET_COLUMN = "Arrival"

# =====================================================
# 2. LOAD DATASET
# =====================================================

print("\n========== LOADING DATASET ==========")

if not DATASET_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found: {DATASET_PATH}"
    )

df = pd.read_excel(
    DATASET_PATH,
    engine="xlrd"
)

df.columns = df.columns.astype(str).str.strip()

# Remove completely empty rows and columns
df = df.dropna(how="all")
df = df.dropna(axis=1, how="all")

print("Dataset Shape:", df.shape)
print("\nAvailable Columns:")
print(df.columns.tolist())

print("\nFirst Five Rows:")
print(df.head())

# =====================================================
# 3. VALIDATE DATASET
# =====================================================

required_columns = FEATURE_COLUMNS + [TARGET_COLUMN]

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing required columns: {missing_columns}"
    )

# =====================================================
# 4. DATA CLEANING
# =====================================================

print("\n========== DATA CLEANING ==========")

for col in CATEGORICAL_FEATURES:
    df[col] = (
        df[col]
        .astype("string")
        .str.strip()
        .fillna("Unknown")
    )

# Standardize month abbreviations
month_mapping = {
    "jan": "Jan",
    "feb": "Feb",
    "mar": "Mar",
    "apr": "Apr",
    "may": "May",
    "jun": "Jun",
    "jul": "Jul",
    "aug": "Aug",
    "sep": "Sep",
    "sept": "Sep",
    "oct": "Oct",
    "nov": "Nov",
    "dec": "Dec"
}

df["Month"] = (
    df["Month"]
    .str.lower()
    .replace(month_mapping)
)

df["Year"] = pd.to_numeric(
    df["Year"],
    errors="coerce"
)

df["Arrival"] = pd.to_numeric(
    df["Arrival"],
    errors="coerce"
)

# Remove records without a valid target
df = df.dropna(subset=["Arrival"])

# Arrival quantities cannot be negative
df = df[df["Arrival"] >= 0].copy()

# Remove duplicate records
df = df.drop_duplicates()

if len(df) < 10:
    raise ValueError(
        "Not enough valid records for model training."
    )

print("Cleaned Dataset Shape:", df.shape)
print("Missing Values:")
print(df.isnull().sum())

print("\nArrival Statistics:")
print(df["Arrival"].describe())

cleaned_path = PROJECT_DIR / "cleaned_dataset.csv"

df.to_csv(cleaned_path, index=False)

print("\nCleaned dataset saved:", cleaned_path)

# =====================================================
# 5. DATASET SUMMARY
# =====================================================

print("\n========== DATASET SUMMARY ==========")

print("Districts:", df["District"].nunique())
print("Blocks:", df["Block"].nunique())
print("Markets:", df["Market"].nunique())
print("Commodities:", df["Commodity"].nunique())

print("\nCommodity Arrival Summary:")

commodity_summary = (
    df.groupby("Commodity", as_index=False)
    ["Arrival"]
    .sum()
    .sort_values("Arrival", ascending=False)
)

print(commodity_summary.head(15))

commodity_summary.to_csv(
    PROJECT_DIR / "commodity_summary.csv",
    index=False
)

market_summary = (
    df.groupby(
        ["District", "Market"],
        as_index=False
    )["Arrival"]
    .sum()
    .sort_values("Arrival", ascending=False)
)

market_summary.to_csv(
    PROJECT_DIR / "market_summary.csv",
    index=False
)

# =====================================================
# 6. DEFINE INPUT AND TARGET
# =====================================================

X = df[FEATURE_COLUMNS].copy()
y = df[TARGET_COLUMN].astype(float)

# Log transformation reduces target skew.
# The original arrival scale is restored after
# prediction using expm1.

y_log = np.log1p(y)

# Random holdout evaluation:
# estimates performance on held-out records,
# NOT future-year forecasting accuracy.

X_train, X_test, y_train, y_test = (
    train_test_split(
        X,
        y_log,
        test_size=TEST_SIZE,
        random_state=RANDOM_STATE
    )
)

# =====================================================
# 7. PREPROCESSING PIPELINE
# =====================================================

print("\n========== PREPROCESSING ==========")

categorical_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            )
        ),
        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)

numerical_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            categorical_transformer,
            CATEGORICAL_FEATURES
        ),
        (
            "numerical",
            numerical_transformer,
            NUMERICAL_FEATURES
        )
    ],
    sparse_threshold=0
)

# =====================================================
# 8. EVALUATION FUNCTION
# =====================================================

def calculate_metrics(actual_log, predicted_log):
    actual = np.expm1(actual_log)

    predicted_log = np.clip(
        predicted_log, -20, 20
    )

    predicted = np.maximum(
        np.expm1(predicted_log), 0
    )

    return {
        "MAE": float(
            mean_absolute_error(actual, predicted)
        ),
        "RMSE": float(
            np.sqrt(
                mean_squared_error(
                    actual, predicted
                )
            )
        ),
        "R2": float(
            r2_score(actual, predicted)
        )
    }


# =====================================================
# 9. MACHINE LEARNING MODELS
# =====================================================

print("\n========== TRAINING ML MODELS ==========")

models = {
    "Ridge Regression": Ridge(
        alpha=1.0
    ),

    "Random Forest": RandomForestRegressor(
        n_estimators=150,
        random_state=RANDOM_STATE,
        n_jobs=-1
    ),

    "Gradient Boosting": GradientBoostingRegressor(
        n_estimators=150,
        learning_rate=0.05,
        random_state=RANDOM_STATE
    ),

    "Extra Trees": ExtraTreesRegressor(
        n_estimators=150,
        random_state=RANDOM_STATE,
        n_jobs=-1
    )
}

results = []
trained_models = {}

for model_name, estimator in models.items():

    print(f"\nTraining: {model_name}")

    pipeline = Pipeline(
        steps=[
            (
                "preprocessor",
                clone(preprocessor)
            ),
            (
                "regressor",
                estimator
            )
        ]
    )

    pipeline.fit(X_train, y_train)

    predictions = pipeline.predict(X_test)

    metrics = calculate_metrics(
        y_test, predictions
    )

    print("MAE:", round(metrics["MAE"], 4))
    print("RMSE:", round(metrics["RMSE"], 4))
    print("R2:", round(metrics["R2"], 4))

    results.append({
        "Model": model_name,
        **metrics
    })

    trained_models[model_name] = pipeline

# =====================================================
# 10. SELECT BEST MACHINE LEARNING MODEL
# =====================================================

comparison_df = pd.DataFrame(results)

comparison_df = comparison_df.sort_values(
    by="RMSE",
    ascending=True
)

print("\n========== MODEL COMPARISON ==========")
print(comparison_df.to_string(index=False))

best_model_name = comparison_df.iloc[0]["Model"]
best_model = trained_models[best_model_name]

print("\nBest ML Model:", best_model_name)

comparison_df.to_csv(
    PROJECT_DIR / "model_comparison.csv",
    index=False
)

# Save best ML pipeline
joblib.dump(
    best_model,
    PROJECT_DIR / "agricultural_model.pkl"
)

# =====================================================
# 11. PREPARE DEEP LEARNING DATA
# =====================================================

print("\n========== DEEP LEARNING ==========")

# Separate preprocessing fit only on training data
nn_preprocessor = clone(preprocessor)

X_train_nn = nn_preprocessor.fit_transform(
    X_train
).astype("float32")

X_test_nn = nn_preprocessor.transform(
    X_test
).astype("float32")

y_train_nn = y_train.to_numpy().astype("float32")
y_test_nn = y_test.to_numpy().astype("float32")

joblib.dump(
    nn_preprocessor,
    PROJECT_DIR / "preprocessor.pkl"
)

input_dimension = X_train_nn.shape[1]

print("Neural Network Input Features:", input_dimension)

# =====================================================
# 12. BUILD NEURAL NETWORK
# =====================================================

neural_model = Sequential([
    Input(shape=(input_dimension,)),

    Dense(
        128,
        activation="relu"
    ),

    Dropout(0.20),

    Dense(
        64,
        activation="relu"
    ),

    Dropout(0.10),

    Dense(
        32,
        activation="relu"
    ),

    Dense(
        1,
        activation="linear"
    )
])

neural_model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="mse",
    metrics=["mae"]
)

neural_model.summary()

# =====================================================
# 13. TRAIN NEURAL NETWORK
# =====================================================

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=15,
    restore_best_weights=True
)

history = neural_model.fit(
    X_train_nn,
    y_train_nn,
    epochs=150,
    batch_size=32,
    validation_split=0.15,
    callbacks=[early_stopping],
    verbose=1
)

# =====================================================
# 14. EVALUATE NEURAL NETWORK
# =====================================================

nn_predictions_log = (
    neural_model.predict(
        X_test_nn,
        verbose=0
    ).flatten()
)

nn_metrics = calculate_metrics(
    y_test_nn,
    nn_predictions_log
)

print("\nNeural Network Performance:")
print("MAE:", nn_metrics["MAE"])
print("RMSE:", nn_metrics["RMSE"])
print("R2:", nn_metrics["R2"])

comparison_df = pd.concat(
    [
        comparison_df,
        pd.DataFrame([{
            "Model": "Neural Network",
            **nn_metrics
        }])
    ],
    ignore_index=True
)

comparison_df = comparison_df.sort_values(
    by="RMSE",
    ascending=True
)

comparison_df.to_csv(
    PROJECT_DIR / "model_comparison.csv",
    index=False
)

# =====================================================
# 15. SAVE H5 MODEL
# =====================================================

h5_path = PROJECT_DIR / "agricultural_model.h5"

neural_model.save(str(h5_path))

print("\nH5 Model Saved:", h5_path)

# =====================================================
# 16. SAVE CONFIG.YAML
# =====================================================

configuration = {
    "project_name":
        "AI-Based Agricultural Commodity Arrival Analysis",

    "dataset": str(DATASET_PATH),

    "target_column": TARGET_COLUMN,

    "categorical_features": CATEGORICAL_FEATURES,

    "numerical_features": NUMERICAL_FEATURES,

    "test_size": TEST_SIZE,

    "random_state": RANDOM_STATE,

    "target_transformation": "log1p",

    "prediction_inverse_transformation": "expm1",

    "models": list(models.keys()) + [
        "Neural Network"
    ],

    "best_ml_model": best_model_name,

    "evaluation_metrics": [
        "MAE",
        "RMSE",
        "R2"
    ],

    "artifacts": {
        "ml_model": "agricultural_model.pkl",
        "neural_model": "agricultural_model.h5",
        "nn_preprocessor": "preprocessor.pkl",
        "metadata": "model_metadata.json"
    }
}

yaml_path = PROJECT_DIR / "config.yaml"

with open(
    yaml_path,
    "w",
    encoding="utf-8"
) as file:
    yaml.safe_dump(
        configuration,
        file,
        sort_keys=False
    )

print("YAML Configuration Saved:", yaml_path)

# =====================================================
# 17. SAVE MODEL METADATA.JSON
# =====================================================

metadata = {
    "project_name":
        "AI-Based Agricultural Commodity Arrival Analysis",

    "created_at": datetime.now(
        timezone.utc
    ).isoformat(),

    "dataset_name": DATASET_PATH.name,

    "total_records": int(len(df)),

    "training_records": int(len(X_train)),

    "testing_records": int(len(X_test)),

    "district_count": int(
        df["District"].nunique()
    ),

    "market_count": int(
        df["Market"].nunique()
    ),

    "commodity_count": int(
        df["Commodity"].nunique()
    ),

    "year_values": sorted([
        int(v)
        for v in df["Year"].dropna().unique()
    ]),

    "features": FEATURE_COLUMNS,

    "target": TARGET_COLUMN,

    "target_unit": "Quintals",

    "best_ml_model": best_model_name,

    "best_ml_metrics": {
        key: float(value)
        for key, value in
        comparison_df[
            comparison_df["Model"] == best_model_name
        ].iloc[0][["MAE", "RMSE", "R2"]].items()
    },

    "neural_network_metrics": nn_metrics,

    "neural_network_epochs": int(
        len(history.history["loss"])
    ),

    "model_comparison": comparison_df.to_dict(
        orient="records"
    ),

    "evaluation_method":
        "Random 80/20 held-out record split",

    "forecasting_limitation":
        "This evaluation does not establish "
        "future-year forecasting accuracy."
}

json_path = PROJECT_DIR / "model_metadata.json"

with open(
    json_path,
    "w",
    encoding="utf-8"
) as file:
    json.dump(
        metadata,
        file,
        indent=4
    )

print("JSON Metadata Saved:", json_path)

# =====================================================
# 18. SAVE TEST PREDICTIONS
# =====================================================

best_predictions_log = best_model.predict(
    X_test
)

actual_arrival = np.expm1(
    y_test.to_numpy()
)

predicted_arrival = np.maximum(
    np.expm1(
        np.clip(best_predictions_log, -20, 20)
    ),
    0
)

prediction_results = X_test.copy()

prediction_results["Actual_Arrival"] = (
    actual_arrival
)

prediction_results["Predicted_Arrival"] = (
    predicted_arrival
)

prediction_results["Absolute_Error"] = (
    np.abs(
        actual_arrival - predicted_arrival
    )
)

prediction_results.to_csv(
    PROJECT_DIR / "test_predictions.csv",
    index=False
)

# =====================================================
# 19. SAVE TRAINING HISTORY
# =====================================================

history_df = pd.DataFrame(
    history.history
)

history_df.to_csv(
    PROJECT_DIR / "training_history.csv",
    index=False
)

# =====================================================
# 20. VERIFY SAVED FILES
# =====================================================

print("\n========== SAVED FILES ==========")

output_files = [
    "agricultural_model.h5",
    "agricultural_model.pkl",
    "preprocessor.pkl",
    "config.yaml",
    "model_metadata.json",
    "cleaned_dataset.csv",
    "commodity_summary.csv",
    "market_summary.csv",
    "model_comparison.csv",
    "test_predictions.csv",
    "training_history.csv"
]

for filename in output_files:
    path = PROJECT_DIR / filename

    if path.exists():
        print("[SAVED]", filename)
    else:
        print("[MISSING]", filename)

print("\n====================================")
print("PROJECT TRAINING COMPLETED")
print("====================================")

print("\nBest ML Model:", best_model_name)
print("\nFinal Model Comparison:")
print(comparison_df.to_string(index=False))

print("\nAll output files are located in:")
print(PROJECT_DIR)
```
