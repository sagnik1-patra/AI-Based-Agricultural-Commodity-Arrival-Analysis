```python

import json
import warnings
from pathlib import Path
from datetime import datetime, timezone

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import yaml

from sklearn.base import clone
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import (
    OneHotEncoder,
    StandardScaler
)
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import (
    train_test_split,
    KFold,
    cross_val_score
)
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings(
    "ignore",
    category=FutureWarning
)

# ==========================================================
# 1. PROJECT CONFIGURATION
# ==========================================================

PROJECT_DIR = Path(
    r"C:\Users\sagni\Downloads\AI-Based Agricultural Commodity Arrival Analysis"
)

DATASET_PATH = (
    PROJECT_DIR / "megamb_data_for_2012_n.xls"
)

PROJECT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

SEED = 42
TEST_SIZE = 0.20
CV_FOLDS = 3

rng = np.random.default_rng(SEED)

CATEGORICAL_FEATURES = [
    "District",
    "Block",
    "Market",
    "Month",
    "Commodity"
]

NUMERICAL_FEATURES = ["Year"]

FEATURE_COLUMNS = (
    CATEGORICAL_FEATURES + NUMERICAL_FEATURES
)

TARGET = "Arrival"

# PSO hyperparameters
N_PARTICLES = 8
N_ITERATIONS = 10

INERTIA_WEIGHT = 0.7
COGNITIVE_COEFFICIENT = 1.5
SOCIAL_COEFFICIENT = 1.5

sns.set_theme(style="whitegrid")

plt.rcParams["figure.figsize"] = (12, 7)
plt.rcParams["font.size"] = 11

# ==========================================================
# 2. LOAD DATASET
# ==========================================================

print("\n========== LOADING DATASET ==========")

if not DATASET_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found: {DATASET_PATH}"
    )

df = pd.read_excel(
    DATASET_PATH,
    engine="xlrd"
)

df.columns = (
    df.columns.astype(str).str.strip()
)

df = df.dropna(how="all")
df = df.dropna(axis=1, how="all")

required_columns = FEATURE_COLUMNS + [TARGET]

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing columns: {missing_columns}"
    )

# ==========================================================
# 3. DATA CLEANING
# Matches the original ML and AIS scripts
# ==========================================================

for col in CATEGORICAL_FEATURES:

    df[col] = (
        df[col]
        .astype("string")
        .str.strip()
        .fillna("Unknown")
    )

month_mapping = {
    "jan": "Jan",
    "feb": "Feb",
    "mar": "Mar",
    "apr": "Apr",
    "may": "May",
    "jun": "Jun",
    "jul": "Jul",
    "aug": "Aug",
    "sep": "Sep",
    "sept": "Sep",
    "oct": "Oct",
    "nov": "Nov",
    "dec": "Dec"
}

df["Month"] = (
    df["Month"]
    .str.lower()
    .replace(month_mapping)
)

df["Year"] = pd.to_numeric(
    df["Year"],
    errors="coerce"
)

df[TARGET] = pd.to_numeric(
    df[TARGET],
    errors="coerce"
)

df = df.dropna(subset=[TARGET])

df = df[
    df[TARGET] >= 0
].copy()

df = df.drop_duplicates()

if len(df) < 20:
    raise ValueError(
        "Not enough valid records for training."
    )

print("Dataset Shape:", df.shape)

print("\nFirst Five Records:")
print(df.head())

df.to_csv(
    PROJECT_DIR / "pso_cleaned_dataset.csv",
    index=False
)

# ==========================================================
# 4. TRAIN / TEST SPLIT
# ==========================================================

X = df[FEATURE_COLUMNS].copy()

# Log transformation for skewed arrival quantities
y = np.log1p(
    df[TARGET].astype(float)
)

X_train, X_test, y_train, y_test = (
    train_test_split(
        X,
        y,
        test_size=TEST_SIZE,
        random_state=SEED
    )
)

# The test set remains outside PSO optimization.

# ==========================================================
# 5. PREPROCESSING
# ==========================================================

categorical_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(
            strategy="most_frequent"
        )
    ),
    (
        "encoder",
        OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False
        )
    )
])

numerical_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(
            strategy="median"
        )
    ),
    (
        "scaler",
        StandardScaler()
    )
])

preprocessor = ColumnTransformer([
    (
        "categorical",
        categorical_pipeline,
        CATEGORICAL_FEATURES
    ),
    (
        "numerical",
        numerical_pipeline,
        NUMERICAL_FEATURES
    )
])

# ==========================================================
# 6. PSO SEARCH SPACE
# ==========================================================

# Each particle contains five Random Forest parameters:
#
# [n_estimators,
#  max_depth,
#  min_samples_split,
#  min_samples_leaf,
#  max_features]

BOUNDS = np.array([
    [50, 250],
    [3, 30],
    [2, 15],
    [1, 8],
    [0.3, 1.0]
], dtype=float)

LOWER_BOUNDS = BOUNDS[:, 0]
UPPER_BOUNDS = BOUNDS[:, 1]

DIMENSIONS = len(BOUNDS)

VELOCITY_LIMIT = (
    UPPER_BOUNDS - LOWER_BOUNDS
) * 0.20


def decode_particle(position):

    position = np.clip(
        position,
        LOWER_BOUNDS,
        UPPER_BOUNDS
    )

    return {
        "n_estimators": int(
            round(position[0])
        ),
        "max_depth": int(
            round(position[1])
        ),
        "min_samples_split": int(
            round(position[2])
        ),
        "min_samples_leaf": int(
            round(position[3])
        ),
        "max_features": float(
            position[4]
        )
    }


def create_pipeline(parameters):

    regressor = RandomForestRegressor(
        **parameters,
        random_state=SEED,
        n_jobs=-1
    )

    return Pipeline([
        (
            "preprocessor",
            clone(preprocessor)
        ),
        (
            "regressor",
            regressor
        )
    ])


# ==========================================================
# 7. PSO FITNESS FUNCTION
# ==========================================================

# Objective: minimize cross-validated RMSE
# in log-transformed arrival space.

fitness_cache = {}

cv = KFold(
    n_splits=CV_FOLDS,
    shuffle=True,
    random_state=SEED
)


def fitness(position):

    parameters = decode_particle(position)

    key = (
        parameters["n_estimators"],
        parameters["max_depth"],
        parameters["min_samples_split"],
        parameters["min_samples_leaf"],
        round(parameters["max_features"], 4)
    )

    if key in fitness_cache:
        return fitness_cache[key]

    model = create_pipeline(parameters)

    scores = cross_val_score(
        model,
        X_train,
        y_train,
        cv=cv,
        scoring="neg_mean_squared_error",
        n_jobs=1
    )

    rmse = float(
        np.sqrt(-scores.mean())
    )

    fitness_cache[key] = rmse

    return rmse


# ==========================================================
# 8. PARTICLE SWARM OPTIMIZATION
# ==========================================================

def run_pso():

    print("\n========== PSO OPTIMIZATION ==========")

    positions = rng.uniform(
        LOWER_BOUNDS,
        UPPER_BOUNDS,
        size=(N_PARTICLES, DIMENSIONS)
    )

    velocities = rng.uniform(
        -VELOCITY_LIMIT,
        VELOCITY_LIMIT,
        size=(N_PARTICLES, DIMENSIONS)
    )

    personal_best_positions = positions.copy()

    personal_best_scores = np.array([
        fitness(position)
        for position in positions
    ])

    best_index = np.argmin(
        personal_best_scores
    )

    global_best_position = (
        personal_best_positions[
            best_index
        ].copy()
    )

    global_best_score = float(
        personal_best_scores[best_index]
    )

    history = []

    for iteration in range(
        1,
        N_ITERATIONS + 1
    ):

        iteration_scores = []

        for particle_index in range(
            N_PARTICLES
        ):

            r1 = rng.random(DIMENSIONS)
            r2 = rng.random(DIMENSIONS)

            cognitive_velocity = (
                COGNITIVE_COEFFICIENT
                * r1
                * (
                    personal_best_positions[
                        particle_index
                    ]
                    - positions[particle_index]
                )
            )

            social_velocity = (
                SOCIAL_COEFFICIENT
                * r2
                * (
                    global_best_position
                    - positions[particle_index]
                )
            )

            velocities[particle_index] = (
                INERTIA_WEIGHT
                * velocities[particle_index]
                + cognitive_velocity
                + social_velocity
            )

            velocities[particle_index] = np.clip(
                velocities[particle_index],
                -VELOCITY_LIMIT,
                VELOCITY_LIMIT
            )

            positions[particle_index] += (
                velocities[particle_index]
            )

            positions[particle_index] = np.clip(
                positions[particle_index],
                LOWER_BOUNDS,
                UPPER_BOUNDS
            )

            current_score = fitness(
                positions[particle_index]
            )

            iteration_scores.append(
                current_score
            )

            # Update personal best
            if (
                current_score
                < personal_best_scores[particle_index]
            ):

                personal_best_scores[
                    particle_index
                ] = current_score

                personal_best_positions[
                    particle_index
                ] = positions[
                    particle_index
                ].copy()

                # Update global best
                if current_score < global_best_score:

                    global_best_score = (
                        current_score
                    )

                    global_best_position = (
                        positions[
                            particle_index
                        ].copy()
                    )

        mean_score = float(
            np.mean(iteration_scores)
        )

        print(
            f"Iteration {iteration}/{N_ITERATIONS}"
            f" | Best CV Log-RMSE: "
            f"{global_best_score:.5f}"
            f" | Mean: {mean_score:.5f}"
        )

        history.append({
            "Iteration": iteration,
            "Global_Best_Log_RMSE":
                global_best_score,
            "Mean_Log_RMSE":
                mean_score,
            "Evaluated_Configurations":
                len(fitness_cache)
        })

    return (
        decode_particle(global_best_position),
        global_best_score,
        pd.DataFrame(history)
    )


best_parameters, best_cv_score, history_df = (
    run_pso()
)

print("\nBest PSO Parameters:")
print(best_parameters)

print("\nBest Cross-Validated Log-RMSE:")
print(best_cv_score)

history_df.to_csv(
    PROJECT_DIR / "pso_optimization_history.csv",
    index=False
)

# ==========================================================
# 9. TRAIN PSO-OPTIMIZED RANDOM FOREST
# ==========================================================

print("\n========== TRAINING PSO MODEL ==========")

pso_model = create_pipeline(
    best_parameters
)

pso_model.fit(
    X_train,
    y_train
)

joblib.dump(
    pso_model,
    PROJECT_DIR / "pso_agricultural_model.pkl"
)

print("Saved: pso_agricultural_model.pkl")

# ==========================================================
# 10. EVALUATION FUNCTIONS
# ==========================================================

def inverse_target(values):

    values = np.asarray(
        values
    ).flatten()

    return np.maximum(
        np.expm1(
            np.clip(values, -20, 20)
        ),
        0
    )


def evaluate(actual, predicted):

    return {
        "MAE": float(
            mean_absolute_error(
                actual,
                predicted
            )
        ),

        "RMSE": float(
            np.sqrt(
                mean_squared_error(
                    actual,
                    predicted
                )
            )
        ),

        "R2": float(
            r2_score(
                actual,
                predicted
            )
        )
    }


# ==========================================================
# 11. EVALUATE PSO MODEL
# ==========================================================

actual = inverse_target(
    y_test.to_numpy()
)

predicted = inverse_target(
    pso_model.predict(X_test)
)

pso_metrics = evaluate(
    actual,
    predicted
)

print("\n========== PSO TEST RESULTS ==========")

for metric, value in pso_metrics.items():
    print(f"{metric}: {value:.5f}")

metrics_df = pd.DataFrame([{
    "Model": "PSO Optimized Random Forest",
    **pso_metrics
}])

metrics_df.to_csv(
    PROJECT_DIR / "pso_evaluation_metrics.csv",
    index=False
)

# ==========================================================
# 12. SAVE PSO RESULT CSV
# ==========================================================

result_df = X_test.copy()

result_df["Actual_Arrival"] = actual

result_df["PSO_Predicted_Arrival"] = predicted

result_df["Absolute_Error"] = np.abs(
    actual - predicted
)

result_df["Squared_Error"] = (
    actual - predicted
) ** 2

result_df.to_csv(
    PROJECT_DIR / "pso_result.csv",
    index=False
)

print("Saved: pso_result.csv")

# ==========================================================
# 13. PSO ACCURACY GRAPH
# ==========================================================

print("\nGenerating PSO Accuracy Graph...")

fig, axes = plt.subplots(
    1, 3,
    figsize=(16, 5)
)

for ax, metric in zip(
    axes,
    ["R2", "MAE", "RMSE"]
):

    value = pso_metrics[metric]

    ax.bar(
        ["PSO Random Forest"],
        [value]
    )

    ax.set_title(
        f"PSO {metric}",
        fontweight="bold"
    )

    ax.set_ylabel(metric)

    ax.text(
        0,
        value,
        f"{value:.4f}",
        ha="center",
        va="bottom"
        if value >= 0 else "top"
    )

fig.suptitle(
    "PSO-Optimized Agricultural Arrival Model Performance",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "pso_accuracy_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ==========================================================
# 14. PSO HEATMAP
# ==========================================================

print("\nGenerating PSO Heatmap...")

top_commodities = (
    df.groupby("Commodity")["Arrival"]
    .sum()
    .nlargest(15)
    .index
)

top_markets = (
    df.groupby("Market")["Arrival"]
    .sum()
    .nlargest(12)
    .index
)

heatmap_df = df[
    df["Commodity"].isin(top_commodities)
    & df["Market"].isin(top_markets)
]

pivot_table = heatmap_df.pivot_table(
    index="Commodity",
    columns="Market",
    values="Arrival",
    aggfunc="sum",
    fill_value=0
)

plt.figure(figsize=(16, 10))

sns.heatmap(
    pivot_table,
    cmap="YlGnBu",
    annot=True,
    fmt=".0f",
    linewidths=0.4,
    cbar_kws={
        "label": "Arrival Quantity (Quintals)"
    }
)

plt.title(
    "PSO Agricultural Commodity Arrival Heatmap",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Market")
plt.ylabel("Commodity")

plt.xticks(
    rotation=45,
    ha="right"
)

plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "pso_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ==========================================================
# 15. COMPARE BASELINE, AIS AND PSO
# ==========================================================

print("\n========== MODEL COMPARISON ==========")

comparison_rows = [{
    "Model": "PSO Optimized Random Forest",
    **pso_metrics
}]

# Baseline model
baseline_path = (
    PROJECT_DIR / "agricultural_model.pkl"
)

if baseline_path.exists():

    baseline_model = joblib.load(
        baseline_path
    )

    baseline_predictions = inverse_target(
        baseline_model.predict(X_test)
    )

    baseline_metrics = evaluate(
        actual,
        baseline_predictions
    )

    comparison_rows.append({
        "Model": "Baseline Best ML",
        **baseline_metrics
    })

# AIS model
ais_path = (
    PROJECT_DIR / "ais_agricultural_model.pkl"
)

if ais_path.exists():

    ais_model = joblib.load(
        ais_path
    )

    ais_predictions = inverse_target(
        ais_model.predict(X_test)
    )

    ais_metrics = evaluate(
        actual,
        ais_predictions
    )

    comparison_rows.append({
        "Model": "AIS Optimized Random Forest",
        **ais_metrics
    })

comparison_df = pd.DataFrame(
    comparison_rows
)

comparison_df.to_csv(
    PROJECT_DIR / "pso_comparison.csv",
    index=False
)

print(comparison_df.to_string(index=False))

fig, axes = plt.subplots(
    1, 3,
    figsize=(19, 6)
)

for ax, metric in zip(
    axes,
    ["MAE", "RMSE", "R2"]
):

    sns.barplot(
        data=comparison_df,
        x=metric,
        y="Model",
        ax=ax
    )

    ax.set_title(
        f"{metric} Comparison",
        fontweight="bold"
    )

    ax.set_ylabel("")

fig.suptitle(
    "Baseline vs AIS vs PSO Model Comparison",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "pso_comparison_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

print("Saved: pso_comparison_graph.png")

# ==========================================================
# 16. PSO RESULT GRAPH
# ==========================================================

print("\nGenerating PSO Result Graph...")

plot_df = result_df.head(
    min(100, len(result_df))
).reset_index(drop=True)

plt.figure(figsize=(16, 7))

plt.plot(
    plot_df.index,
    plot_df["Actual_Arrival"],
    marker="o",
    markersize=3,
    label="Actual Arrival"
)

plt.plot(
    plot_df.index,
    plot_df["PSO_Predicted_Arrival"],
    marker="s",
    markersize=3,
    label="PSO Predicted Arrival"
)

plt.title(
    "PSO Actual vs Predicted Agricultural Arrivals",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Test Sample Index")
plt.ylabel("Arrival (Quintals)")

plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "pso_result_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ==========================================================
# 17. PSO PREDICTION CSV
# ==========================================================

print("\nGenerating PSO Prediction CSV...")

all_features = df[FEATURE_COLUMNS].copy()

all_predictions = inverse_target(
    pso_model.predict(
        all_features
    )
)

prediction_df = all_features.copy()

prediction_df["Actual_Arrival"] = (
    df["Arrival"].to_numpy()
)

prediction_df["PSO_Predicted_Arrival"] = (
    all_predictions
)

prediction_df["Absolute_Error"] = np.abs(
    prediction_df["Actual_Arrival"]
    - prediction_df["PSO_Predicted_Arrival"]
)

prediction_df.to_csv(
    PROJECT_DIR / "pso_prediction.csv",
    index=False
)

print("Saved: pso_prediction.csv")

# ==========================================================
# 18. PSO PREDICTION GRAPH
# ==========================================================

print("\nGenerating PSO Prediction Graph...")

commodity_prediction = (
    prediction_df.groupby(
        "Commodity",
        as_index=False
    )[
        [
            "Actual_Arrival",
            "PSO_Predicted_Arrival"
        ]
    ]
    .sum()
    .sort_values(
        "Actual_Arrival",
        ascending=False
    )
    .head(15)
    .set_index("Commodity")
)

commodity_prediction.plot(
    kind="bar",
    figsize=(16, 7),
    width=0.8
)

plt.title(
    "PSO Commodity-wise Actual vs Predicted Arrivals",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Commodity")
plt.ylabel("Total Arrival (Quintals)")

plt.xticks(
    rotation=45,
    ha="right"
)

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "pso_prediction_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ==========================================================
# 19. PSO OPTIMIZATION CONVERGENCE GRAPH
# ==========================================================

print("\nGenerating PSO Optimization Graph...")

plt.figure(figsize=(12, 6))

plt.plot(
    history_df["Iteration"],
    history_df["Global_Best_Log_RMSE"],
    marker="o",
    linewidth=2,
    label="Global Best Fitness"
)

plt.plot(
    history_df["Iteration"],
    history_df["Mean_Log_RMSE"],
    marker="s",
    linestyle="--",
    label="Mean Particle Fitness"
)

plt.title(
    "Particle Swarm Optimization Convergence",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Iteration")
plt.ylabel("Cross-Validated Log-RMSE")

plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "pso_optimization_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ==========================================================
# 20. SAVE PSO YAML CONFIGURATION
# ==========================================================

config = {
    "project_name":
        "AI-Based Agricultural Commodity Arrival Analysis",

    "optimization_algorithm":
        "Particle Swarm Optimization",

    "dataset": str(DATASET_PATH),

    "target": TARGET,

    "target_unit": "Quintals",

    "features": FEATURE_COLUMNS,

    "random_state": SEED,

    "test_size": TEST_SIZE,

    "cross_validation_folds": CV_FOLDS,

    "number_of_particles": N_PARTICLES,

    "number_of_iterations": N_ITERATIONS,

    "inertia_weight": INERTIA_WEIGHT,

    "cognitive_coefficient":
        COGNITIVE_COEFFICIENT,

    "social_coefficient":
        SOCIAL_COEFFICIENT,

    "fitness_metric":
        "Cross-validated log-space RMSE",

    "best_hyperparameters":
        best_parameters,

    "best_cv_log_rmse":
        float(best_cv_score),

    "model_file":
        "pso_agricultural_model.pkl"
}

with open(
    PROJECT_DIR / "pso_config.yaml",
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        config,
        file,
        sort_keys=False
    )

print("Saved: pso_config.yaml")

# ==========================================================
# 21. SAVE PSO MODEL METADATA
# ==========================================================

metadata = {
    "project_name":
        "AI-Based Agricultural Commodity Arrival Analysis",

    "algorithm":
        "Particle Swarm Optimization",

    "created_at":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "dataset_records": int(
        len(df)
    ),

    "training_records": int(
        len(X_train)
    ),

    "testing_records": int(
        len(X_test)
    ),

    "features": FEATURE_COLUMNS,

    "target": TARGET,

    "target_unit": "Quintals",

    "best_hyperparameters":
        best_parameters,

    "best_cv_log_rmse":
        float(best_cv_score),

    "test_metrics":
        pso_metrics,

    "comparison_results":
        comparison_df.to_dict(
            orient="records"
        ),

    "evaluation_method":
        "Held-out random 80/20 split",

    "note":
        "PSO optimizes Random Forest hyperparameters "
        "using cross-validation on training data. "
        "The test set is used for final evaluation."
}

with open(
    PROJECT_DIR / "pso_model_metadata.json",
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        metadata,
        file,
        indent=4
    )

print("Saved: pso_model_metadata.json")

# ==========================================================
# 22. VERIFY OUTPUT FILES
# ==========================================================

print("\n========== PSO OUTPUT FILES ==========")

output_files = [
    "pso_agricultural_model.pkl",
    "pso_config.yaml",
    "pso_model_metadata.json",
    "pso_cleaned_dataset.csv",
    "pso_accuracy_graph.png",
    "pso_heatmap.png",
    "pso_comparison.csv",
    "pso_comparison_graph.png",
    "pso_result.csv",
    "pso_result_graph.png",
    "pso_prediction.csv",
    "pso_prediction_graph.png",
    "pso_optimization_graph.png",
    "pso_optimization_history.csv",
    "pso_evaluation_metrics.csv"
]

for filename in output_files:

    path = PROJECT_DIR / filename

    if path.exists():
        print("[SAVED]", filename)
    else:
        print("[MISSING]", filename)

print("\n========================================")
print("PSO OPTIMIZATION COMPLETED SUCCESSFULLY")
print("========================================")

print("\nBest PSO Parameters:")
print(best_parameters)

print("\nPSO Test Metrics:")
print(pso_metrics)

print("\nProject Directory:")
print(PROJECT_DIR)
```
